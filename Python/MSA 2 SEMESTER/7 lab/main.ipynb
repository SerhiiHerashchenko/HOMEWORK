# Багатовимірне метричне шкалювання (ММШ)
**Класична модель Торгерсона**

Цей блокнот містить детальне рішення Завдання 16 для варіантів 4 та 9. Використовуються виключно теоретичні формули з класичної моделі (без застосування готових бібліотек на кшталт `scikit-learn` для ММШ).

Усі операції зведені до базових матричних обчислень за допомогою `numpy`.

```python
import numpy as np
import matplotlib.pyplot as plt

# Налаштування для зручного виводу матриць
np.set_printoptions(precision=4, suppress=True)

def mds_torgerson(X_init, variant_name):
    print(f"{'='*50}\nРІШЕННЯ ДЛЯ {variant_name.upper()}\n{'='*50}")
    n = len(X_init)
    print("Початкові координати об'єктів (X_init):\n", X_init, "\n")

    # ---------------------------------------------------------
    # Крок 0. Стандартизація вихідних даних (Зауваження 9.3)
    # z_ij = (x_ij - mean_j) / sigma_j
    # ---------------------------------------------------------
    means = np.mean(X_init, axis=0)
    stds = np.std(X_init, axis=0)

    # Уникаємо ділення на нуль, якщо дисперсія нульова
    stds[stds == 0] = 1

    Z = (X_init - means) / stds

    print("Середні значення за осями (x, y):", means)
    print("Середньоквадратичні відхилення (sigma_x, sigma_y):", stds)
    print("0. Стандартизована матриця об'єктів Z:\n", Z, "\n")

    # ---------------------------------------------------------
    # Крок 1. Розрахунок матриці відстаней γ за стандартизованими даними
    # ---------------------------------------------------------
    gamma = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            gamma[i, j] = np.sqrt(np.sum((Z[i] - Z[j])**2))
    print("1. Матриця відстаней γ:\n", gamma, "\n")

    # ---------------------------------------------------------
    # Крок 2. Побудова матриці B (Формула 9.5)
    # ---------------------------------------------------------
    B = np.zeros((n, n))
    gamma_sq = gamma**2
    sum_i = np.sum(gamma_sq, axis=0)
    sum_j = np.sum(gamma_sq, axis=1)
    sum_all = np.sum(gamma_sq)

    for i in range(n):
        for j in range(n):
            B[i, j] = 0.5 * (-gamma_sq[i, j] + sum_j[i]/n + sum_i[j]/n - sum_all/(n**2))
    print("2. Матриця попарних скалярних добутків B:\n", B, "\n")

    # ---------------------------------------------------------
    # Крок 3. Власні числа та вектори (Спектральний розклад)
    # ---------------------------------------------------------
    eigenvalues, eigenvectors = np.linalg.eigh(B)

    # Сортування за спаданням
    idx = np.argsort(eigenvalues)[::-1]
    eigenvalues = eigenvalues[idx]
    eigenvectors = eigenvectors[:, idx]

    print("3. Власні числа λ:\n", eigenvalues, "\n")
    # print("Власні вектори l (стовпці):\n", eigenvectors, "\n")

    p = np.sum(eigenvalues > 1e-8)
    print(f"Кількість додатних власних чисел (p) = {p}. Розмірність нового простору = {p}.\n")

    # ---------------------------------------------------------
    # Крок 4. Обчислення координат X у шкальному просторі
    # ---------------------------------------------------------
    lambda_p = eigenvalues[:p]
    V_p = eigenvectors[:, :p]

    Lambda_half = np.diag(np.sqrt(lambda_p))
    X_new = V_p @ Lambda_half

    print("4. Матриця координат X (результат ММШ):\n", X_new, "\n")

    # ---------------------------------------------------------
    # Крок 5. Графічне відображення (3 графіки для наочності)
    # ---------------------------------------------------------
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

    # Графік 1: Початкові дані
    ax1.scatter(X_init[:, 0], X_init[:, 1], c='blue', edgecolors='black', s=100, zorder=3)
    ax1.set_title("Початкове розташування")
    ax1.grid(True, linestyle='--', alpha=0.7)
    ax1.set_aspect('equal')

    # Графік 2: Після стандартизації (Z)
    ax2.scatter(Z[:, 0], Z[:, 1], c='green', edgecolors='black', s=100, zorder=3)
    ax2.set_title("Після стандартизації (Z-оцінка)")
    ax2.grid(True, linestyle='--', alpha=0.7)
    ax2.set_aspect('equal')
    ax2.axhline(0, color='black', linewidth=1)
    ax2.axvline(0, color='black', linewidth=1)

    # Графік 3: Після ММШ
    ax3.scatter(X_new[:, 0], X_new[:, 1], c='red', edgecolors='black', s=100, zorder=3)
    ax3.set_title("Після ММШ")
    ax3.grid(True, linestyle='--', alpha=0.7)
    ax3.set_aspect('equal')
    ax3.axhline(0, color='black', linewidth=1)
    ax3.axvline(0, color='black', linewidth=1)

    plt.suptitle(f"Трансформація об'єктів для {variant_name}", fontsize=14, y=1.05)
    plt.tight_layout()
    plt.show()

    return X_new
```

### Виконання для Варіанту 4
Зчитані координати з графіка (x, y).

```python
# Координати об'єктів для Варіанту 4
V4_points = np.array([
    [2, 2], [2, 3], [3, 3], [3, 4], [3, 5], [4, 3], [5, 3]
], dtype=float)

X_new_v4 = mds_torgerson(V4_points, "Варіанту 4")
```
