```python
import numpy as np
import pandas as pd

# ==========================================
# ЯЧЕЙКА 1: Ввод исходных данных (Вариант 4)
# ==========================================
# Объекты: 31-40. Признаки: 1, 6, 8, 9
# ЗАМЕНИТЕ НУЛИ НА РЕАЛЬНЫЕ ДАННЫЕ ИЗ ТАБЛИЦЫ!
data = np.array([
    [3524.0, 13.0, 99.0, 0.57], # Объект 31
    [4300.0, 11.0, 99.0, 0.4], # Объект 32
    [2800.0, 32.0, 68.0, 4.8], # Объект 33
    [43900.0, 35.0, 76.0, 2.6], # Объект 34
    [45000.0, 447.0, 96.0, 1.0], # Объект 35
    [23100.0, 189.0, 99.0, 1.83], # Объект 36
    [38600.0, 123.0, 99.0, 0.3], # Объект 37
    [10500.0, 108.0, 85.0, 0.36], # Объект 38
    [149200.0, 8.8, 99.0, 0.2], # Объект 39
    [18000.0, 7.7, 62.0, 3.2]  # Объект 40
])

# Названия для таблиц
countries = ["Новая Зеландия","Норвегия","ОАЭ","ПАР","Южная Корея","Северная Корея","Польша","Португалия","Россия","Саудовская Аравия"]
features = ["X1", "X6", "X8", "X9"]

print("Данные для варианта 4 загружены. Размерность:", data.shape)
```

```python
from scipy.stats import chi2

# ==========================================
# ЯЧЕЙКА 2. Нормалізація даних і перевірка значущості матриці R
# ==========================================
n, m = data.shape

# Центрируем и нормируем данные
Z = (data - np.mean(data, axis=0)) / np.std(data, axis=0, ddof=1)

print("--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Стандартизованные данные Z (в полном объеме) ---")
display(pd.DataFrame(Z, index=countries, columns=features).round(4))

# Корреляционная матрица R
R = np.corrcoef(Z, rowvar=False)

print("\n--- 1. Кореляційна матриця R ---")
display(pd.DataFrame(R, index=features, columns=features).round(4))

# Тест Бартлетта на сферичность
det_R = np.linalg.det(R)
chi2_spher = -(n - 1 - (2 * m + 5) / 6) * np.log(det_R)
df_spher = m * (m - 1) / 2
alpha = 0.1
chi2_kr_spher = chi2.ppf(1 - alpha, df_spher)

print("\n--- Перевірка матриці R на значущість (Сферичність) ---")
print(f"Визначник матриці R: {det_R:.6f}")
print(f"Статистика chi^2_st = {chi2_spher:.4f}")
print(f"Критичне значення chi^2_kr = {chi2_kr_spher:.4f} (при df={df_spher}, alpha={alpha})")

if chi2_spher > chi2_kr_spher:
    print("Висновок: матриця R значущо відрізняється від одиничної. Факторний аналіз доцільний.")
else:
    print("Висновок: ознаки слабко корелюють між собою.")
```

```python
# ==========================================
# ЯЧЕЙКА 3. Редукована матриця Rh, матриця A та Тест Бартлетта (Алгоритм 8.2)
# ==========================================
np.fill_diagonal(R, 0)
communalities_initial = np.max(np.abs(R), axis=1)
np.fill_diagonal(R, 1)

print("--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Початкові оцінки спільностей ---")
print(pd.Series(communalities_initial, index=features).round(4).to_string())

Rh = R.copy()
np.fill_diagonal(Rh, communalities_initial)

print("\n--- 2. Редукована кореляційна матриця Rh ---")
display(pd.DataFrame(Rh, index=features, columns=features).round(4))

# Спектральное разложение
eigenvalues, eigenvectors = np.linalg.eigh(Rh)
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# ---> ДОБАВЛЕНО: Вывод собственных значений по убыванию <---
print("\n--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Власні числа матриці Rh (Лямбда) ---")
print(pd.Series(eigenvalues, index=[f"lambda_{i+1}" for i in range(m)]).round(4).to_string())

print("\n--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Власні вектори матриці Rh (V) ---")
display(pd.DataFrame(eigenvectors, index=features, columns=[f"v_{i+1}" for i in range(m)]).round(4))

r_factors = min(3, m)
print(f"\nЗа умовою завдання виділяємо факторів: r = {r_factors}")

# Матрица A
Lambda = np.diag(np.sqrt(np.maximum(eigenvalues[:r_factors], 0)))
V = eigenvectors[:, :r_factors]
A = V @ Lambda

factor_names = [f"f_{i+1}" for i in range(r_factors)]
print("\n--- 3. Матриця факторних навантажень A (до обертання) ---")
display(pd.DataFrame(A, index=features, columns=factor_names).round(4))

# АЛГОРИТМ 8.2
h2_reprod = np.sum(A**2, axis=1)
D2 = np.diag(1 - h2_reprod)
R_reprod = A @ A.T + D2

print("\n--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Матриця D^2 (діагональ характерностей) ---")
display(pd.DataFrame(D2, index=features, columns=features).round(4))

print("\n--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Відтворена матриця кореляцій (R_reprod) ---")
display(pd.DataFrame(R_reprod, index=features, columns=features).round(4))

det_R_reprod = np.linalg.det(R_reprod)
chi2_st_alg = (n - (2 * m + 5) / 6 - (2 * r_factors) / 3) * np.log(det_R_reprod / det_R)
df_alg = 0.5 * ((m - r_factors)**2 - m - r_factors)

print("\n--- Алгоритм 8.2: Тест на достатність кількості факторів ---")
print(f"Статистика chi^2_st = {chi2_st_alg:.4f}")
print(f"Ступені свободи df = {df_alg}")

df_alg = abs(df_alg)
if df_alg > 0:
    chi2_kr_alg = chi2.ppf(1 - alpha, df_alg)
    print(f"Критичне значення chi^2_kr = {chi2_kr_alg:.4f} (alpha={alpha})")
    if chi2_st_alg > chi2_kr_alg:
        print("Висновок: chi^2_st > chi^2_kr. Кількість факторів НЕ достатня, треба збільшити r.")
    else:
        print("Висновок: chi^2_st <= chi^2_kr. Виділеної кількості факторів достатньо.")
else:
    print("Висновок: Ступені свободи <= 0. Математично критерій не застосовується (модель перенасичена).")
```

```python
# ==========================================
# ЯЧЕЙКА 4. Аналітичне обертання Квартимакс-критерієм
# ==========================================
def quartimax_rotation_verbose(A_init, tol=1e-6, max_iter=100):
    m_vars, r_facs = A_init.shape
    A_rot = A_init.copy()
    T_total = np.eye(r_facs)

    print("--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Деталі обертання (Ітерація 1) ---")

    for iteration in range(max_iter):
        max_angle = 0
        for k in range(r_facs - 1):
            for l in range(k + 1, r_facs):
                ak = A_rot[:, k]
                al = A_rot[:, l]

                u = ak**2 - al**2
                v = 2 * ak * al

                num = 2 * np.sum(u * v)
                den = np.sum(u**2 - v**2)
                phi = 0.25 * np.arctan2(num, den)

                if iteration == 0:
                    print(f"Фактори {k+1} та {l+1}: num = {num:.4f}, den = {den:.4f} => Кут phi = {np.degrees(phi):.4f}°")

                if abs(phi) > max_angle:
                    max_angle = abs(phi)

                c, s = np.cos(phi), np.sin(phi)
                T = np.eye(r_facs)
                T[k, k] = c; T[k, l] = s
                T[l, k] = -s; T[l, l] = c

                A_rot = A_rot @ T
                T_total = T_total @ T

        if max_angle < tol:
            print(f"\nОбертання зійшлося за {iteration+1} ітерацій.")
            break

    return A_rot, T_total

if r_factors > 1:
    A_rotated, T_matrix = quartimax_rotation_verbose(A)
    print("\n--- 4. Матриця перетворення T (повна) ---")
    display(pd.DataFrame(T_matrix).round(4))
else:
    A_rotated = A.copy()
    print("Обертання не застосовується.")

print("\n--- Матриця факторних навантажень A^H (ПІСЛЯ обертання Квартимакс) ---")
df_A_rot = pd.DataFrame(A_rotated, index=features, columns=factor_names)
display(df_A_rot.round(4))
```

```python
# ==========================================
# ЯЧЕЙКА 5. Розрахунок матриці значень факторів на об'єктах
# ==========================================
R_inv = np.linalg.inv(R)

print("--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Обернена матриця R^(-1) ---")
display(pd.DataFrame(R_inv, index=features, columns=features).round(4))

F = Z @ R_inv @ A_rotated

print("\n--- 5. Матриця F (значення факторів для кожного об'єкта) ---")
df_F = pd.DataFrame(F, index=countries, columns=factor_names)
display(df_F.round(4))
```

```python
# ==========================================
# ЯЧЕЙКА 6. Розподіл дисперсії та спільності
# ==========================================
A_sq = A_rotated**2

print("--- [ПРОМЕЖУТОЧНЫЙ ШАГ]: Матриця квадратів навантажень (A^H)^2 ---")
display(pd.DataFrame(A_sq, index=features, columns=factor_names).round(4))

h_squared = np.sum(A_sq, axis=1)
d_squared = 1 - h_squared

df_variance = pd.DataFrame(A_sq, index=features, columns=[f"Внесок {fac}^2" for fac in factor_names])
df_variance['Спільність h^2'] = h_squared
df_variance['Специфічність d^2'] = d_squared

print("\n--- 6. Внески факторів у дисперсію ознак ---")
display(df_variance.round(4))

factor_variances = np.sum(A_sq, axis=0)
total_communality = np.sum(h_squared)

df_total_contrib = pd.DataFrame({
    'Дисперсія фактора (v_j)': factor_variances,
    'Внесок до сумарної спільності (%)': (factor_variances / total_communality) * 100
}, index=factor_names)

print("\n--- Частки дисперсій факторів ---")
print(f"Сумарна спільність (v_0): {total_communality:.4f}")
print(f"Сумарна характерна дисперсія: {np.sum(d_squared):.4f}\n")
display(df_total_contrib.round(2))
```
