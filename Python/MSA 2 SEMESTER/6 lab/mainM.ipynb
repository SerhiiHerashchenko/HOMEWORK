```python
import numpy as np
import pandas as pd

# ==========================================
# ЯЧЕЙКА 1: Ввод исходных данных (Вариант 9)
# ==========================================
# Объекты: 13-22. Признаки: 6, 7, 8, 9
data = np.array([
    [14641.0, 5494.0, 94.0, 77.0], # Объект 13
    [122.0, 47.0, 12.0, 24.0], # Объект 14
    [748.0, 57.0, 44.0, 48.0], # Объект 15
    [573.0, 11.0, 42.0, 73.0], # Объект 16
    [275.0, 283.0, 26.0, 52.0], # Объект 17
    [12170.0, 51.0, 57.0, 98.0], # Объект 18
    [13047.0, 77.0, 78.0, 95.0], # Объект 19
    [17500.0, 188.0, 69.0, 97.0], # Объект 20
    [19904.0, 2.8, 77.0, 97.0], # Объект 21
    [377.0, 124.0, 26.0, 78.0]  # Объект 22
])

# Названия для таблиц
countries = ["Гонконг","Ефиопия","Египет","Замбия","Индия","Ирландия","Испания","Италия","Канада","Китай"]
features = ["X6", "X7", "X8", "X9"]

print("Данные для варианта 9 загружены. Размерность:", data.shape)
```

```python
# ==========================================
# ЯЧЕЙКА 2. Нормализация данных и расчет R
# ==========================================
# Центрируем и нормируем данные (Z-масштабирование)
Z = (data - np.mean(data, axis=0)) / np.std(data, axis=0, ddof=1)

# Корреляционная матрица R
R = np.corrcoef(Z, rowvar=False)

print("--- 1. Корреляционная матрица признаков R ---")
display(pd.DataFrame(R, index=features, columns=features).round(3))
```

```python
# ==========================================
# ЯЧЕЙКА 3. Редуцированная корреляционная матрица Rh
# ==========================================
# Рассчитываем общности через квадрат множественной корреляции (SMC)
# SMC_i = 1 - 1 / R_inv_ii
R_inv = np.linalg.inv(R)
communalities = 1 - 1 / np.diag(R_inv)

# Копируем R и заменяем диагональ на общности
Rh = R.copy()
np.fill_diagonal(Rh, communalities)

print("--- 2. Редуцированная матрица Rh (на диагонали - общности) ---")
display(pd.DataFrame(Rh, index=features, columns=features).round(3))
```

```python
# ==========================================
# ЯЧЕЙКА 4. Спектральное разложение и матрица A
# ==========================================
# Находим собственные числа и векторы матрицы Rh
eigenvalues, eigenvectors = np.linalg.eigh(Rh)

# Сортируем по убыванию
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print("--- Собственные числа матрицы Rh ---")
print(np.round(eigenvalues, 3))

# Берем первые m факторов (максимум 3, как просят в задании, но исключаем отрицательные корни)
m = 3
valid_m = min(m, sum(eigenvalues > 0))
Lambda = np.diag(np.sqrt(eigenvalues[:valid_m]))
V = eigenvectors[:, :valid_m]

# Матрица факторных нагрузок A
A = V @ Lambda
factor_names = [f"Фактор {i+1}" for i in range(valid_m)]

print(f"\n--- 3-4. Матрица факторных нагрузок A (для {valid_m} факторов) ---")
df_A = pd.DataFrame(A, index=features, columns=factor_names)
display(df_A.round(3))
```

```python
# ==========================================
# ЯЧЕЙКА 5. Матрица F значений факторов на объектах
# ==========================================
# Вычисляем регрессионным методом: F = Z * R^(-1) * A
F = Z @ R_inv @ A

print("--- 5. Матрица значений факторов на объектах F ---")
df_F = pd.DataFrame(F, index=countries, columns=factor_names)
display(df_F.round(3))
```

```python
# ==========================================
# ЯЧЕЙКА 6. Вклады факторов в дисперсию
# ==========================================
# Дисперсия каждого фактора (сумма квадратов факторных нагрузок в столбце матрицы A)
factor_variances = np.sum(A**2, axis=0)

# Общая общность (сумма диагональных элементов Rh)
total_communality = np.sum(communalities)

# Относительные вклады в %
relative_contributions = (factor_variances / total_communality) * 100

print("--- 6. Относительные вклады факторов в суммарную дисперсию ---")
df_contrib = pd.DataFrame({
    'Дисперсия фактора': factor_variances,
    'Доля от объясненной дисперсии (%)': relative_contributions
}, index=factor_names)
display(df_contrib.round(2))
```
